```python
from sentence_transformers import SentenceTransformer
import numpy as np
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
news_stream = [
    "Stock market reaches all-time high.",
    "Heavy rainfall causes flooding in the city.",
    "Artificial Intelligence is transforming healthcare.",
    "India wins the cricket championship.",
    "Scientists discover a new exoplanet."
]
```

```python
vector_db = []
```

```python
for idx, news in enumerate(news_stream):

    embedding = model.encode(news)

    vector_db.append({
        "id": idx,
        "text": news,
        "vector": embedding
    })

    print(f"Indexed News {idx+1}:")
    print(news)
    print("Vector Dimension:", len(embedding))
    print("-"*50)
```

```python
query = "AI in medicine"

query_vector = model.encode(query)
```

```python
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

best_score = -1
best_text = ""

for item in vector_db:

    score = cosine_similarity(query_vector, item["vector"])

    if score > best_score:
        best_score = score
        best_text = item["text"]

print("Query :", query)
print("Most Similar News :", best_text)
print("Similarity Score :", round(best_score,4))
```
